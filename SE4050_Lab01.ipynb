```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D

%matplotlib inline
```

```
exp_array = np.random.exponential(scale=1.0, size=(4, 4))
print(exp_array)
```

```
exp_samples = np.random.exponential(scale=1.0, size=100000)
uni_samples = np.random.rand(100000)
norm_samples = np.random.randn(100000)

plt.hist(exp_samples, density=True, bins=100, histtype='step', color='red', label='Exponential')
plt.hist(uni_samples, density=True, bins=100, histtype='step', color='blue', label='Uniform')
plt.hist(norm_samples, density=True, bins=100, histtype='step', color='green', label='Normal')

plt.xlim(-1, 6)
plt.ylim(0, 1.2)
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Comparison of Distributions')
plt.legend(loc='upper right')
plt.grid(alpha=0.3)
plt.show()
```

```
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax.set_title('Z = X² + Y²')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()
```

```
df = pd.read_csv("Pokemon.csv")  # The file must be in the same folder as your notebook

features = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']

pearson_corr = df[features].corr(method='pearson')
spearman_corr = df[features].corr(method='spearman')

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

sns.heatmap(pearson_corr, annot=True, fmt='.2f', cmap='coolwarm',
            vmin=-1, vmax=1, square=True, ax=axes[0])
axes[0].set_title('Pearson Correlation')

sns.heatmap(spearman_corr, annot=True, fmt='.2f', cmap='coolwarm',
            vmin=-1, vmax=1, square=True, ax=axes[1])
axes[1].set_title('Spearman Rank Correlation')

plt.tight_layout()
plt.show()

print("\nPearson Correlation Matrix:")
print(pearson_corr)
print("\nSpearman Correlation Matrix:")
print(spearman_corr)

```

```
df_au = pd.read_csv('Au_nanoparticle_dataset.csv')
keep_cols = ['N_total', 'N_bulk', 'N_surface', 'R_avg']
new_df = df_au[keep_cols].copy()


filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("Dataset shape (rows, columns):", df.shape)
print("\nFirst 5 rows:")
print(df.head())
```

```

print(df.columns.tolist())
```

```
new_df = df[['N_total', 'N_bulk', 'N_surface', 'R_avg']]

print("New DataFrame shape:", new_df.shape)
print("\nFirst 5 rows of new DataFrame:")
print(new_df.head())
```

```
print("First 20 samples:")
print(new_df.head(20))
```

```
# All statistics at once
stats = new_df.describe()
print("Statistical values:")
print(stats)

# Calculate separately
print("\nMean values:")
print(new_df.mean())

print("\nStandard Deviation values:")
print(new_df.std())

print("\nQuartile values (25%, 50%, 75%):")
print(new_df.quantile([0.25, 0.5, 0.75]))
```

```
# 1x4 layout
fig, axes = plt.subplots(1, 4, figsize=(20, 5))

# Create histogram for each feature
for i, column in enumerate(new_df.columns):
    axes[i].hist(new_df[column], bins=30, color='blue', alpha=0.7, edgecolor='black')
    axes[i].set_title(f'Histogram of {column}', fontsize=12)
    axes[i].set_xlabel('Value', fontsize=10)
    axes[i].set_ylabel('Frequency', fontsize=10)
    axes[i].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

```

```

# Create pairplot
plt.figure(figsize=(12, 10))
sns.pairplot(new_df, diag_kind='hist', corner=False)
plt.suptitle('Pairplot of All Features', y=1.02, fontsize=14)
plt.show()
```

```

g = sns.PairGrid(new_df)
g.map_upper(sns.histplot)           # bivariate histogram
g.map_diag(sns.histplot, kde=True)  # histogram with KDE overlay
g.map_lower(sns.kdeplot)            # bivariate kernel density estimate
plt.show()
```
